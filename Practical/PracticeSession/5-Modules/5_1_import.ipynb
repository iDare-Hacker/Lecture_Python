##### Importing Modules in Python: Modules and Packages
In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4])
```

```python
from math import *
print(sqrt(16))
print(pi)
```

```python
import os

# Step 1: Create package folder
os.makedirs('package', exist_ok=True)
```

```python
%%writefile package/maths.py
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b
```

```python
%%writefile package/__init__.py
# This file marks this folder as a Python package
# You can leave it empty or use it to expose some parts of the module
```

```python
from package.maths import addition, subtraction

print("Addition:", addition(5, 3))
print("Subtraction:", subtraction(5, 3))
```

```python
from package import maths
maths.addition(2,3)
```

#### Conclusion
Importing modules and packages in Python allows you to organize your code, reuse functionalities, and keep your projects clean and manageable. By understanding how to import modules, specific functions, and use relative imports within packages, you can structure your Python applications more effectively.
